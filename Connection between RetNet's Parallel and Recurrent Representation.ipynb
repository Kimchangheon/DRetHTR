## Connection Between RetNet's Parallel and Recurrent Representations

To better understand the connection between the parallel and recurrent forms, consider the following matrix operations.

### 1. Defining $Q$, $K^\top$, and $V$

$$
\begin{aligned}
Q &=
\begin{bmatrix}
q_1 \\
q_2 \\
q_3
\end{bmatrix},
\qquad
K^\top =
\begin{bmatrix}
k_1^\top & k_2^\top & k_3^\top
\end{bmatrix},
\qquad
V =
\begin{bmatrix}
v_1 \\
v_2 \\
v_3
\end{bmatrix}.
\end{aligned}
$$

### 2. Computing $QK^\top$ and defining $D$

$$
QK^\top =
\begin{bmatrix}
q_1 k_1^\top & q_1 k_2^\top & q_1 k_3^\top \\
q_2 k_1^\top & q_2 k_2^\top & q_2 k_3^\top \\
q_3 k_1^\top & q_3 k_2^\top & q_3 k_3^\top
\end{bmatrix},
\qquad
D =
\begin{bmatrix}
1 & 0 & 0 \\
\gamma & 1 & 0 \\
\gamma^2 & \gamma & 1
\end{bmatrix}.
$$

### 3. Computing the retention matrix

$$
(QK^\top)\odot D =
\begin{bmatrix}
q_1 k_1^\top & 0 & 0 \\
\gamma q_2 k_1^\top & q_2 k_2^\top & 0 \\
\gamma^2 q_3 k_1^\top & \gamma q_3 k_2^\top & q_3 k_3^\top
\end{bmatrix}.
$$

### 4. Final retention operation

$$
Y =
\begin{bmatrix}
q_1 (k_1^\top v_1) \\
q_2 (\gamma k_1^\top v_1 + k_2^\top v_2) \\
q_3 (\gamma^2 k_1^\top v_1 + \gamma k_2^\top v_2 + k_3^\top v_3)
\end{bmatrix}
=
\begin{bmatrix}
q_1 S_1 \\
q_2 S_2 \\
q_3 S_3
\end{bmatrix},
$$

$$
\text{where } S_n = \gamma S_{n-1} + k_n^\top v_n.
$$

This formulation shows how the parallel retention computation can be rewritten as a recurrent state update, demonstrating the validity and flexibility of the retention mechanism.
